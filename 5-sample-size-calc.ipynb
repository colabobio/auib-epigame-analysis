## Sample size calculations

### Sample size calculation using all responses

```python
# This is the expected reduction in quarantine between groups due to barrier in G2
g2_factor = 0.25
```

```python
users_g1 = users[users['group'] == 1]
users_g2 = users[users['group'] == 2]
print(len(users_g1), len(users_g2))
```

```python
mean = users_g1['quarantine'].mean()
std = users_g1['quarantine'].std()
delta = mean - g2_factor * mean

# Formula 2 in 
# https://faculty.washington.edu/jelmore/articles_online/Wittes-Sample_size_calc.pdf
N = 2 * std * ((1.96 + 0.84)**2)/ delta**2 

print('Group size =', round(N))
```

### Sample size calculation using only non-zero responses

```python
users_nzq = users[users['quarantine'] > 0]
users_nzq_g1 = users_nzq[users_nzq['group'] == 1]
users_nzq_g2 = users_nzq[users_nzq['group'] == 2]
print(len(users_nzq_g1), len(users_nzq_g2))
```

```python
mean = users_nzq_g1['quarantine'].mean()
std = users_nzq_g1['quarantine'].std()
delta = mean - g2_factor * mean
N = 2 * std * ((1.96 + 0.84)**2)/ delta**2 
print('Group size =', round(N))
```
